### 2. Excel 파일 읽기

`pandas` 라이브러리를 사용하여 업로드된 Excel 파일을 데이터프레임으로 읽습니다. 파일 이름이 `your_file.xlsx`라고 가정합니다.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 여기에 업로드한 Excel 파일의 이름을 입력하세요.
file_name = 'MCI 갯수.xlsx'

try:
    # Excel 파일을 데이터프레임으로 읽기 (첫 번째 시트 사용)
    df = pd.read_excel(file_name)
    print(f"{file_name} 파일이 성공적으로 로드되었습니다.")

    # 데이터프레임의 처음 5행을 표시하여 데이터 구조 확인
    print("\n데이터프레임 미리보기:")
    display(df.head())

    # 데이터프레임 정보 요약
    print("\n데이터프레임 정보:")
    df.info()

except FileNotFoundError:
    print(f"오류: '{file_name}' 파일을 찾을 수 없습니다. 파일 이름을 확인하거나 파일이 올바르게 업로드되었는지 확인하세요.")
except Exception as e:
    print(f"파일을 읽는 중 오류가 발생했습니다: {e}")
```

### 5. Matplotlib 한글 폰트 설정 적용

런타임 재시작 후 이 셀을 실행하여 Matplotlib에 나눔고딕 폰트를 적용합니다.

```python
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# 설치된 나눔고딕 폰트 경로 설정
font_path = '/usr/share/fonts/truetype/nanum/NanumGothic.ttf'
font_name = fm.FontProperties(fname=font_path, size=10).get_name()
plt.rc('font', family=font_name)

# 마이너스 부호 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False

print(f"Matplotlib에 '{font_name}' 폰트가 적용되었습니다.")
```

```python
# 꺾은선 그래프의 x축과 y축으로 사용할 열 이름을 지정하세요.
# 예: 'Date' 및 'Sales'와 같이 실제 열 이름으로 변경해야 합니다.
X_COLUMN_NAME = '년도' # 예시, 실제 열 이름으로 변경
Y_COLUMN_NAME = 'MCI 비율' # 예시, 실제 열 이름으로 변경 (수정됨)

if X_COLUMN_NAME in df.columns and Y_COLUMN_NAME in df.columns:
    # x축 컬럼이 날짜/시간 형식이라면 변환합니다.
    # df[X_COLUMN_NAME] = pd.to_datetime(df[X_COLUMN_NAME], errors='coerce')

    # 결측값이 있는 행을 제거 (필요한 경우)
    plot_df = df.dropna(subset=[X_COLUMN_NAME, Y_COLUMN_NAME])

    plt.figure(figsize=(12, 6))
    sns.lineplot(x=X_COLUMN_NAME, y=Y_COLUMN_NAME, data=plot_df)
    plt.title(f'{Y_COLUMN_NAME}의 {X_COLUMN_NAME}에 따른 변화')
    plt.xlabel(X_COLUMN_NAME)
    plt.ylabel(Y_COLUMN_NAME)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.xticks(rotation=45) # x축 레이블이 겹치지 않도록 회전
    plt.tight_layout() # 레이아웃 조정
    plt.show()
else:
    print(f"오류: 지정된 열 '{X_COLUMN_NAME}' 또는 '{Y_COLUMN_NAME}'이(가) 데이터프레임에 존재하지 않습니다. 열 이름을 확인해주세요.")
```
