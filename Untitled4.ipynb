```python
# ==========================================
# 1. CSV 파일 불러오기
# ==========================================

from google.colab import files
import pandas as pd
import numpy as np
from scipy.stats import skew

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("파일명:", file_name)
print("데이터 크기:", df.shape)

display(df)
```

```python
# ==========================================
# 2. 토픽별 문서 수와 비중 계산
# ==========================================

# 전체 문서 수
total_documents = df["document_count"].sum()

# 토픽별 결과
topic_result = df[
    ["topic", "분야", "document_count"]
].copy()

# 전체 문서에서 차지하는 비중(%)
topic_result["비중(%)"] = (
    topic_result["document_count"] / total_documents * 100
)

# 문서 수가 많은 순서로 정렬
topic_result = topic_result.sort_values(
    "document_count",
    ascending=False
).reset_index(drop=True)

print(f"전체 문서 수: {total_documents:,}편")

display(topic_result)
```

```python
# ==========================================
# 3. 집중도 지표 계산
# ==========================================

# 지니계수 함수
def gini_coefficient(values):
    values = np.asarray(values, dtype=float)
    values = values[~np.isnan(values)]
    values = np.sort(values)

    n = len(values)

    if n == 0 or np.sum(values) == 0:
        return np.nan

    return (
        2 * np.sum(np.arange(1, n + 1) * values)
        / (n * np.sum(values))
        - (n + 1) / n
    )


# ==========================================
# 전체 토픽 기준
# ==========================================

document_counts = df["document_count"].astype(float).values

# ① 지니계수
gini = gini_coefficient(document_counts)

# ② 왜도
skewness = skew(document_counts, bias=False)

# ③ HHI
shares = document_counts / document_counts.sum()
hhi = np.sum(shares ** 2)

# ④ 가장 큰 토픽
max_index = df["document_count"].idxmax()

max_topic = df.loc[max_index, "topic"]
max_topic_name = df.loc[max_index, "분야"]
max_count = df.loc[max_index, "document_count"]
max_share = max_count / total_documents * 100


# ==========================================
# 결과 출력
# ==========================================

print("======================================")
print("       전체 문헌 주제 집중도 분석")
print("======================================")

print(f"전체 문서 수              : {total_documents:,}편")
print(f"토픽 수                   : {len(df)}개")

print()
print(f"가장 많은 토픽            : Topic {max_topic}")
print(f"분야                      : {max_topic_name}")
print(f"해당 토픽 문서 수         : {max_count:,}편")
print(f"전체 문서 중 비중         : {max_share:.2f}%")

print()
print(f"지니계수                  : {gini:.4f}")
print(f"HHI                       : {hhi:.4f}")
print(f"왜도(쏠림지수)            : {skewness:.4f}")
```

```python
# ==========================================
# 4. Outlier 제외 후 주제 집중도
# ==========================================

# Topic -1 제외
topic_df = df[df["topic"] != -1].copy()

# Outlier 제외한 문서 수
topic_total = topic_df["document_count"].sum()

# 각 실제 토픽의 비중
topic_df["주제 내 비중(%)"] = (
    topic_df["document_count"] / topic_total * 100
)

# 문서 수
topic_counts = topic_df["document_count"].astype(float).values

# 지니계수
topic_gini = gini_coefficient(topic_counts)

# 왜도
topic_skewness = skew(topic_counts, bias=False)

# HHI
topic_shares = topic_counts / topic_counts.sum()
topic_hhi = np.sum(topic_shares ** 2)

# 가장 큰 실제 토픽
max_idx = topic_df["document_count"].idxmax()

largest_topic = topic_df.loc[max_idx, "topic"]
largest_topic_name = topic_df.loc[max_idx, "분야"]
largest_count = topic_df.loc[max_idx, "document_count"]
largest_share = (
    largest_count / topic_total * 100
)


print("======================================")
print("      Outlier 제외 주제 집중도 분석")
print("======================================")

print(f"전체 문서 수                  : {total_documents:,}편")
print(f"Outlier 문서 수               : {total_documents - topic_total:,}편")
print(f"실제 토픽 문서 수             : {topic_total:,}편")

print()
print(f"가장 큰 토픽                  : Topic {largest_topic}")
print(f"분야                          : {largest_topic_name}")
print(f"해당 토픽 문서 수             : {largest_count:,}편")
print(f"실제 토픽 문서 중 비중        : {largest_share:.2f}%")

print()
print(f"지니계수                      : {topic_gini:.4f}")
print(f"HHI                           : {topic_hhi:.4f}")
print(f"왜도(쏠림지수)                : {topic_skewness:.4f}")
```

```python
# ==========================================
# 6. 집중도 지표 결과표
# ==========================================

concentration_result = pd.DataFrame({
    "지표": [
        "전체 문서 수",
        "토픽 수",
        "최대 토픽",
        "최대 토픽 문서 수",
        "최대 토픽 비중(전체 문서 기준)",
        "지니계수(전체 토픽)",
        "HHI(전체 토픽)",
        "왜도(전체 토픽)",
        "지니계수(Outlier 제외)",
        "HHI(Outlier 제외)",
        "왜도(Outlier 제외)"
    ],

    "값": [
        total_documents,
        len(df),
        f"Topic {max_topic} ({max_topic_name})",
        max_count,
        round(max_share, 2),
        round(gini, 4),
        round(hhi, 4),
        round(skewness, 4),
        round(topic_gini, 4),
        round(topic_hhi, 4),
        round(topic_skewness, 4)
    ]
})

display(concentration_result)
```
